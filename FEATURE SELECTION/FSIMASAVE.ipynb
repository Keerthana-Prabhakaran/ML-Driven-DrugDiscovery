```python
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('SSFS2_Features.csv')
df = pd.DataFrame(data)

fig = plt.figure(figsize=(10, 10))   # CREATE FIGURE FIRST

X = list(df.iloc[:, 1])
Y = list(df.iloc[:, 2])

plt.bar(X, Y, color='c')
plt.title(" Substructure Feature", fontweight="bold")
plt.xlabel("Features", fontweight='bold')
plt.ylabel("Scores", fontweight='bold')
plt.xticks(rotation=80, fontweight="bold")
plt.yticks(fontweight="bold")
plt.savefig("SS_IMA.jpg", bbox_inches='tight')  # SAVE AFTER PLOTTING
plt.show()
```
